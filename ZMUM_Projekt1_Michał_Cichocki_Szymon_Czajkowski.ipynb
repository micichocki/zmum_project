# Uzupełnianie danych

## Sposób 1: Mediana

```python
int_columns = [
  'blueWins', 'blueFirstBlood', 'blueTotalExperience', 'blueGoldDiff',
  'blueExperienceDiff','redFirstBlood', 'redTotalExperience', 'redGoldDiff',
  'redExperienceDiff', 'blueKillsDiff', 'blueAssistsDiff', 'blueWardsDiff',
  'blueTowersDiff', 'blueTotalMinionsKilledDiff', 'blueEliteMonstersDiff',
  'blueJungleMonstersDiff', 'blueDragonsDiff', 'blueHeraldsDiff'
]

float_columns = list(df_empty.columns.difference(int_columns))
```

```python
df_empty_median = df_empty.copy()

for column in df_empty_median.columns:
    median_val = df_empty_median[column].median()
    df_empty_median[column] = df_empty_median[column].fillna(median_val)
    if column in int_columns:
        df_empty_median[column] = df_empty_median[column].astype(int)
df_imputed_median = df_empty_median

np.sum(df_imputed_median.isnull(), axis=0)
```

## Sposób 2: Średnia

```python
df_empty_mean = df_empty.copy()

for column in df_empty_mean.columns:
    mean_val = df_empty_mean[column].mean()
    df_empty_mean[column] = df_empty_mean[column].fillna(mean_val)
    if column in int_columns:
        df_empty_mean[column] = df_empty_mean[column].round().astype(int)
df_imputed_mean = df_empty_mean

np.sum(df_imputed_mean.isnull(), axis=0)
```

## Sposób 3: Regresja

```python
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler

df_empty_regression = df_empty.copy()

scaler = StandardScaler()
df_scaled = pd.DataFrame(
    scaler.fit_transform(df_empty_regression),
    columns=df_empty_regression.columns
)

imputer = IterativeImputer(
    estimator=LinearRegression(),
    max_iter=10,
    initial_strategy="mean",
    random_state=123
)

df_imputed_scaled = pd.DataFrame(
    imputer.fit_transform(df_scaled),
    columns=df_scaled.columns
)

df_imputed_regression = pd.DataFrame(
        scaler.inverse_transform(df_imputed_scaled),
    columns=df_empty_regression.columns
)

for column in df_imputed_regression.columns:
    if column in int_columns:
        df_imputed_regression[column] = df_imputed_regression[column].round().astype(int)

print(np.sum(df_imputed_regression.isna(), axis=0))
```

# Skalowanie cech

## Skalowanie min-max

```python
min_scale = 0
max_scale = 1

def min_max_scale(data):
  scaled_data = data.copy()

  for column in scaled_data.columns:
    if column == 'blueWins':
        continue
    min = scaled_data[column].min()
    max = scaled_data[column].max()
    scaled_data[column] = ((scaled_data[column] - min) / (max - min)) * (max_scale - min_scale) + min_scale

  return scaled_data
```

## Standaryzacja

```python
def standarize(data):
    scaled_data = data.copy()

    for column in scaled_data.columns:
        if column == 'blueWins':
            continue
        mean = scaled_data[column].mean()
        std = scaled_data[column].std()
        scaled_data[column] = (scaled_data[column] - mean) / std

    return scaled_data
```

# Podział danych na zbiór testowy i treningowy

```python
def split(X, Y, test_size=0.2, random_state=None):
    if random_state is not None:
        np.random.seed(random_state)

    data = pd.concat([X, Y], axis=1)
    label_name = 'blueWins'

    train_parts = []
    test_parts = []

    for label, group in data.groupby(label_name):
        n_test = int(len(group) * test_size)
        shuffled = group.sample(frac=1, random_state=random_state)

        test_parts.append(shuffled.iloc[:n_test])
        train_parts.append(shuffled.iloc[n_test:])

    train_df = pd.concat(train_parts).sample(frac=1, random_state=random_state)
    test_df = pd.concat(test_parts).sample(frac=1, random_state=random_state)

    X_train = train_df.drop(columns=[label_name])
    Y_train = train_df[label_name]
    X_test = test_df.drop(columns=[label_name])
    Y_test = test_df[label_name]

    return X_train, X_test, Y_train, Y_test
```

# Trening i ewaluacja modeli

## Funkcje do pomiaru metryk

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

### ACC

```python
def acc(cm):
    ((tp, fn), (fp, tn)) = cm
    return (tp + tn) / (tp + tn + fp + fn)
```

## F1

```python
def f1(cm, c):
    ((tp, fn), (fp, tn)) = cm
    sum = 0.0
    for i in range(c):
        sum += (2 * tp) / (2 * tp + fp + fn)

    return sum / c
```

## Walidacja krzyżowa

```python
def cross_validation(model, X, y, k=5):
    indices = np.arange(len(X))
    np.random.shuffle(indices)

    fold_sizes = np.full(k, len(X) // k)
    fold_sizes[:len(X) % k] += 1

    curr = 0
    acc_scores = []
    f1_scores = []

    for fold_size in fold_sizes:
        start, stop = curr, curr + fold_size
        test_idx = indices[start:stop]
        train_idx = np.concatenate([indices[:start], indices[stop:]])

        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

        curr = stop

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)
        cm = confusion_matrix(y_test, y_pred)

        acc_scores.append(acc(cm))
        f1_scores.append(f1(cm, 2))

    return { 'acc': np.mean(acc_scores), 'f1': np.mean(f1_scores) }
```

```python
from sklearn.metrics import accuracy_score, f1_score
from sklearn.model_selection import RandomizedSearchCV


def optimize_svc(X, Y, n_iter=50):

    X_train, X_test, Y_train, Y_test = split(
        X,
        Y,
        random_state=RANDOM_STATE,
    )

    param_distributions = [
        {
            'C': [0.01, 0.1, 1, 10, 100, 1000],
            'kernel': ['rbf', 'sigmoid'],
            'gamma': ['scale', 'auto', 0.1, 0.01, 0.001],
            'class_weight': [None, 'balanced'],
            'random_state': [RANDOM_STATE]
        },
        {
            'C': [0.01, 0.1, 1, 10, 100],
            'kernel': ['linear'],
            'class_weight': [None, 'balanced'],
            'random_state': [RANDOM_STATE]
        },
        {
            'C': [0.1, 1, 10, 100],
            'kernel': ['poly'],
            'degree': [2, 3, 4],
            'gamma': ['scale', 'auto'],
            'coef0': [0.0, 0.1, 0.5],
            'class_weight': [None, 'balanced'],
            'random_state': [RANDOM_STATE]
        }
    ]

    scoring = {
        'f1': 'f1',
        'accuracy': 'accuracy'
    }

    svc = SVC()

    rs = RandomizedSearchCV(
        estimator=svc,
        param_distributions=param_distributions,
        n_iter=n_iter,
        scoring=scoring,
        refit='f1',
        cv=5,
        n_jobs=-1,
        verbose=1,
        random_state=RANDOM_STATE
    )

    rs.fit(X_train, Y_train)

    print(f"Najlepsze parametry: {rs.best_params_}")

    return rs.best_estimator_, rs.best_params_, X_test, Y_test


best_svc_model_grid_search, best_params, X_test, Y_test = optimize_svc(
    X_imputed_pycaret_scaled,
    Y_imputed_pycaret_scaled,
    n_iter=25
)

y_pred_test = best_svc_model_grid_search.predict(X_test)
final_acc_grid_search = accuracy_score(Y_test, y_pred_test)
final_f1_grid_search = f1_score(Y_test, y_pred_test)

print(f"Final acc GridSearch: {final_acc_grid_search:.4f}")
print(f"Final f1 GridSearch:  {final_f1_grid_search:.4f}")

optimization_results.append({
    'Metoda': 'Grid Search',
    'Accuracy': final_acc_grid_search,
    'F1 Score': final_f1_grid_search
})
```

## Optuna

```python
import optuna

def optimize_svc_optuna(X, Y, n_trials=50):
    X_train, X_test, Y_train, Y_test = split(
        X,
        Y,
        random_state=RANDOM_STATE,
    )

    def objective(trial):
        kernel = trial.suggest_categorical('kernel', ['linear', 'rbf', 'poly', 'sigmoid'])
        c = trial.suggest_float('C', 1e-2, 1e3, log=True)
        class_weight = trial.suggest_categorical('class_weight', [None, 'balanced'])

        degree = 3
        gamma = 'scale'
        coef0 = 0.0

        if kernel == 'poly':
            degree = trial.suggest_int('degree', 2, 5)
            gamma_mode = trial.suggest_categorical('gamma_mode', ['str', 'float'])
            if gamma_mode == 'str':
                gamma = trial.suggest_categorical('gamma_str', ['scale', 'auto'])
            else:
                gamma = trial.suggest_float('gamma_float', 1e-4, 1e-1, log=True)
            coef0 = trial.suggest_float('coef0', 0.0, 1.0)

        elif kernel in ['rbf', 'sigmoid']:
            gamma_mode = trial.suggest_categorical('gamma_mode_rbf', ['str', 'float'])
            if gamma_mode == 'str':
                gamma = trial.suggest_categorical('gamma_str_rbf', ['scale', 'auto'])
            else:
                gamma = trial.suggest_float('gamma_float_rbf', 1e-4, 1e-1, log=True)

            if kernel == 'sigmoid':
                 coef0 = trial.suggest_float('coef0_sig', 0.0, 1.0)

        clf = SVC(
            C=c,
            kernel=kernel,
            degree=degree,
            gamma=gamma,
            coef0=coef0,
            class_weight=class_weight,
            random_state=RANDOM_STATE
        )

        cv_results = cross_validation(clf, X_train, Y_train, k=5)

        return cv_results['f1']

    study = optuna.create_study(direction='maximize')
    optuna.logging.set_verbosity(optuna.logging.WARNING)

    study.optimize(objective, n_trials=n_trials, show_progress_bar=True)

    bp = study.best_params

    final_params = {
        'C': bp['C'],
        'kernel': bp['kernel'],
        'class_weight': bp['class_weight'],
        'random_state': RANDOM_STATE
    }

    if bp['kernel'] == 'poly':
        final_params['degree'] = bp['degree']
        final_params['coef0'] = bp['coef0']
        if bp['gamma_mode'] == 'str':
            final_params['gamma'] = bp['gamma_str']
        else:
            final_params['gamma'] = bp['gamma_float']

    elif bp['kernel'] in ['rbf', 'sigmoid']:
        if bp['gamma_mode_rbf'] == 'str':
            final_params['gamma'] = bp['gamma_str_rbf']
        else:
            final_params['gamma'] = bp['gamma_float_rbf']

        if bp['kernel'] == 'sigmoid':
             final_params['coef0'] = bp['coef0_sig']

    print(f"Najlepsze F1 z walidacji: {study.best_value:.4f}")
    print("Trenowanie finalnego modelu...")

    best_model = SVC(**final_params)
    best_model.fit(X_train, Y_train)

    return best_model, final_params, X_test, Y_test



best_svc_model_optuna, best_params_optuna, X_test, Y_test = optimize_svc_optuna(X_imputed_pycaret_scaled, Y_imputed_pycaret_scaled, n_trials=20)

y_pred_test = best_svc_model_optuna.predict(X_test)
final_acc_optuna = accuracy_score(Y_test, y_pred_test)
final_f1_optuna = f1_score(Y_test, y_pred_test)

print("Final acc Optuna:", final_acc_optuna)
print("Final f1 Optuna: ", final_f1_optuna)

optimization_results.append({
    'Metoda': 'Optuna',
    'Accuracy': final_acc_optuna,
    'F1 Score': final_f1_optuna
})
```

## Algorytm genetyczny

```python
import random
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import f1_score

POPULATION_SIZE = 20
GENERATIONS = 10
TOURNAMENT_SIZE = 3
MUTATION_RATE = 0.2
ELITISM_COUNT = 2

GENE_SPACE = {
    'C': [0.01, 0.1, 1, 5, 10, 50, 100, 500, 1000],
    'kernel': ['linear', 'rbf', 'poly', 'sigmoid'],
    'gamma': ['scale', 'auto', 0.001, 0.01, 0.1, 1],
    'degree': [2, 3, 4, 5],
    'coef0': [0.0, 0.1, 0.5, 1.0, 2.0],
    'class_weight': [None, 'balanced']
}

def create_individual():
    return {
        'C': random.choice(GENE_SPACE['C']),
        'kernel': random.choice(GENE_SPACE['kernel']),
        'gamma': random.choice(GENE_SPACE['gamma']),
        'degree': random.choice(GENE_SPACE['degree']),
        'coef0': random.choice(GENE_SPACE['coef0']),
        'class_weight': random.choice(GENE_SPACE['class_weight'])
    }

def get_fitness(individual, X, Y):
    clf = SVC(
        C=individual['C'],
        kernel=individual['kernel'],
        gamma=individual['gamma'],
        degree=individual['degree'],
        coef0=individual['coef0'],
        class_weight=individual['class_weight'],
        random_state=RANDOM_STATE
    )

    cv_results = cross_validation(clf, X, Y, k=5)
    return cv_results['f1']

def selection(population, fitness_scores):
    candidates_indices = random.sample(range(len(population)), TOURNAMENT_SIZE)
    best_idx = candidates_indices[0]
    best_fitness = fitness_scores[best_idx]

    for idx in candidates_indices[1:]:
        if fitness_scores[idx] > best_fitness:
            best_fitness = fitness_scores[idx]
            best_idx = idx

    return population[best_idx]

def crossover(parent1, parent2):
    child = {}
    for key in parent1.keys():
        if random.random() < 0.5:
            child[key] = parent1[key]
        else:
            child[key] = parent2[key]
    return child

def mutate(individual):
    mutated = individual.copy()
    for key in GENE_SPACE.keys():
        if random.random() < MUTATION_RATE:
            mutated[key] = random.choice(GENE_SPACE[key])
    return mutated


def run_genetic_algorithm(X, Y):
    X_train, X_test, Y_train, Y_test = split(
        X,
        Y,
        random_state=RANDOM_STATE,
    )

    population = [create_individual() for _ in range(POPULATION_SIZE)]

    best_global_fitness = 0.0
    best_global_params = None

    for gen in range(GENERATIONS):
        fitness_scores = []
        for ind in population:
            score = get_fitness(ind, X_train, Y_train)
            fitness_scores.append(score)

        max_fitness_idx = np.argmax(fitness_scores)
        current_best_fitness = fitness_scores[max_fitness_idx]
        current_best_params = population[max_fitness_idx]

        if current_best_fitness > best_global_fitness:
            best_global_fitness = current_best_fitness
            best_global_params = current_best_params
            print(f"{best_global_fitness:.4f} | Params: {best_global_params}")
        else:
            print(f"[Gen {gen+1}] Max F1: {current_best_fitness:.4f}")

        new_population = []

        sorted_indices = np.argsort(fitness_scores)[::-1]
        for i in range(ELITISM_COUNT):
            elite_idx = sorted_indices[i]
            new_population.append(population[elite_idx])

        while len(new_population) < POPULATION_SIZE:
            parent1 = selection(population, fitness_scores)
            parent2 = selection(population, fitness_scores)

            child = crossover(parent1, parent2)
            child = mutate(child)

            new_population.append(child)

        population = new_population


    final_model = SVC(
        C=best_global_params['C'],
        kernel=best_global_params['kernel'],
        gamma=best_global_params['gamma'],
        degree=best_global_params['degree'],
        coef0=best_global_params['coef0'],
        class_weight=best_global_params['class_weight'],
        random_state=RANDOM_STATE
    )
    final_model.fit(X_train, Y_train)

    y_pred = final_model.predict(X_test)

    return final_model, accuracy_score(Y_test, y_pred), f1_score(Y_test, y_pred)


best_model_ga, final_acc_ga, final_f1_ga = run_genetic_algorithm(X_imputed_pycaret_scaled, Y_imputed_pycaret_scaled)

print(f"Accuracy: {final_acc_ga:.4f}")
print(f"F1 Score: {final_f1_ga:.4f}")

optimization_results.append({
    'Metoda': 'Genetic Algorithm',
    'Accuracy': final_acc_ga,
    'F1 Score': final_f1_ga
})
```
